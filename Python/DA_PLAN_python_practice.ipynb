```python
def check(mark):
  if mark>=50:
    return("PASS")
  else:
    return("FAIL")
total=0
marks = [50,40,70,88,98]
for mark in marks :
  total=total+mark
  print(mark,check(mark))
avg=total/len(marks)
print("AVERAGE:",avg)
print("TOTAL:",total)
```
