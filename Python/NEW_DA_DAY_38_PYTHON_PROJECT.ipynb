Because mean() alone gives 1 value per category.

When to use print() and when df

This is where many beginners confuse.

Use df

When you want pretty table display.

Example:

df

or

df.head()

Shows table nicely.

Best for:

dataframe
rows
columns
analysis

Example:

df["Sales"]

Output:

0 50000
1 700
2 3500
Use print()

When showing messages or multiple outputs.

Example:

print("Total Sales:", df["Sales"].sum())

Output:

Total Sales: 86200

Good for numbers/text.

Example:

print(df["Sales"].sum())

Good.

Rule
Table/dataframe → df
Text/numbers/messages → print()
