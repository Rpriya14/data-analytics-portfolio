```python
import pandas as pd

data = {
    'Month':['Jan','Feb','Mar','Apr','May'],
    'Sales':[100,120,90,140,160],
    'Profit':[20,25,15,30,35]
}

df=pd.DataFrame(data)
print(df)
```

MATPLOTLIB

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(df["Month"],df["Sales"])
```

```python
plt.plot(df["Month"],df["Sales"],marker="o")
plt.title("MONTHLY SALES")
plt.xlabel("MONTH")
plt.ylabel("SALES")
plt.show()
```

```python
plt.bar(df["Month"],df["Sales"])
```

```python
plt.bar(df["Month"],df["Sales"])
plt.title("SALES COMPARISON")
plt.xlabel("MONTH")
plt.ylabel("SALES")
```

```python
plt.pie(df['Profit'],labels=df['Month'],autopct='%1.1f%%')
plt.show()
```

HISTOGRAM
range based

Question histogram asks:

👉 How are numbers distributed?

Group numeric values into ranges.

```python
plt.hist(df["Sales"])
plt.show()
```

```python
counts, bins, patches = plt.hist(df["Sales"])
print("Counts:", counts)
print("Bins:", bins)
plt.show()
```

SCATTER PLOT
Relationship bet 2 nos

```python
plt.scatter(df['Sales'],df['Profit'])
plt.show()
```

SEABORN

```python
import seaborn as sns
```

```python
sns.lineplot(x='Month',y='Sales',data=df)
plt.show()
```

```python
sns.barplot(x='Month',y='Sales',data=df)
plt.show()
```

```python
sns.scatterplot(x='Sales',y='Profit',data=df)
plt.show()
```

HEATMAP
